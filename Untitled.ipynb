```python
import pandas as pd
from IPython.display import display

# Lê o arquivo único com todos os resultados
df = pd.read_csv('resultados.csv')

# Define as listas de algoritmos para separar as tabelas
algoritmos_rapidos = [
    'Arrays.sort(Java)', 'Merge Sort (Clássico)', 
    'Quick Sort (Simples)', 'Quick Sort (Shuffle)', 
    'Counting Sort'
]
algoritmos_lentos = [
    'Insertion Sort', 'Selection Sort (Simples)', 
    'Selection Sort (Estável)', 'Bubble Sort (Otimizado)', 
    'Bubble Sort (Simples)'
]

# Cria as ordens de colunas base
ordem_cenarios = ['Ordenado', 'Aleatório', 'Ordem inversa', 'Parcialmente ordenado (15%)']
ordem_tamanhos_rapidos = ['n=20000', 'n=50000', 'n=100000']
ordem_tamanhos_lentos = ['n=20000', 'n=50000', 'n=100000']

# Estilos CSS comuns para ambas as tabelas
estilos = [
    {'selector': 'th', 'props': [('text-align', 'center')]},
    {'selector': 'caption', 'props': [('caption-side', 'top'), ('font-size', '16px'), ('font-weight', 'bold'), ('color', 'black'), ('margin-bottom', '15px')]},
    {'selector': 'th:nth-child(5), td:nth-child(5)', 'props': [('border-left', '2px solid #bbbbbb')]},
    {'selector': 'th:nth-child(8), td:nth-child(8)', 'props': [('border-left', '2px solid #bbbbbb')]}
]

df_rapidos = df[df['Algoritmo'].isin(algoritmos_rapidos)]
tabela_rapidos = pd.pivot_table(df_rapidos, values='Tempo_ms', index='Algoritmo', columns=['Cenario', 'Tamanho'], aggfunc='mean')
tabela_rapidos = tabela_rapidos.reindex(columns=pd.MultiIndex.from_product([ordem_cenarios, ordem_tamanhos_rapidos]))
tabela_rapidos = tabela_rapidos.fillna('-')

tabela_estilizada_rapidos = tabela_rapidos.style.set_caption("Tabela 1: Algoritmos Eficientes e Lineares (Até 300k)") \
                                                .set_properties(**{'text-align': 'center'}) \
                                                .set_table_styles(estilos)

df_lentos = df[df['Algoritmo'].isin(algoritmos_lentos)]
tabela_lentos = pd.pivot_table(df_lentos, values='Tempo_ms', index='Algoritmo', columns=['Cenario', 'Tamanho'], aggfunc='mean')
tabela_lentos = tabela_lentos.reindex(columns=pd.MultiIndex.from_product([ordem_cenarios, ordem_tamanhos_lentos]))
tabela_lentos = tabela_lentos.fillna('-')

tabela_estilizada_lentos = tabela_lentos.style.set_caption("Tabela 2: Algoritmos Quadráticos O(n²) (Até 300k)") \
                                              .set_properties(**{'text-align': 'center'}) \
                                              .set_table_styles(estilos)

# Exibe as duas tabelas renderizadas na mesma célula
display(tabela_estilizada_rapidos)
print("\n") # Dá um pequeno espaçamento em branco entre as duas tabelas
display(tabela_estilizada_lentos)
```
